```python
# ===== DATA DIRI =====
nama = "Ervan Ramdani"
asal_sekolah = "SMKN 1 Cimahi"
jurusan = "Teknik Otomasi Industri"
alamat = "Bandung"
pekerjaan = "Mahasiswa Teknik Informatika (Kelas Karyawan)"

hobi = ["Belajar Teknologi Baru", "Otomotif & Kelistrikan", "Membaca Buku"]
cita_cita = "Pengusaha Sukses"

# ===== BIODATA (dictionary) =====
biodata = {
    "nama": nama,
    "sekolah": asal_sekolah,
    "jurusan": jurusan,
    "alamat": alamat,
    "pekerjaan": pekerjaan,
    "hobi": hobi,
    "cita_cita": cita_cita,
}

# ===== FUNGSI PERKENALAN =====
def perkenalan(data):
    print("=" * 45)
    print("            PERKENALAN DIRI")
    print("=" * 45)
    print(f"Halo, nama saya {data['nama']}.")
    print(f"Saya merupakan lulusan dari {data['sekolah']}, jurusan {data['jurusan']}.")
    print(f"Saya tinggal di {data['alamat']} dan saat ini menempuh pendidikan {data['pekerjaan']}.")
    print("\nHobi saya:")
    for i, h in enumerate(data["hobi"], start=1):
        print(f"  {i}. {h}")
    print(f"\nCita-cita saya menjadi {data['cita_cita']}.")
    print("Terima kasih sudah membaca!")
    print("=" * 45)

# ===== JALANKAN =====
perkenalan(biodata)
```
